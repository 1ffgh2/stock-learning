## 核心概念：DataFrame = Python 里的 Excel 表格

```python
import pandas as pd
print("pandas 版本:", pd.__version__)
```

## 2. 手工创建 DataFrame

```python
data = {
    "日期":     ["7-01", "7-02", "7-03", "7-04", "7-05"],
    "股票":     ["茅台", "茅台", "茅台", "茅台", "茅台"],
    "开盘价":   [1800, 1812, 1808, 1820, 1830],
    "收盘价":   [1812, 1808, 1820, 1830, 1825],
    "成交量_万": [520, 480, 610, 550, 490]
}

df = pd.DataFrame(data)
df
```

## 3. 五大核心操作（以后每天都用）

```python
# ① 看形状（多少行，多少列）
df.shape
```

```python
# ② 取某一列
df["收盘价"]
```

```python
# ③ 统计摘要（平均值、最大最小、标准差...）
df.describe()
```

```python
# ④ 筛选符合条件的行
df[df["收盘价"] > 1820]
```

```python
# ⑤ 排序
df.sort_values("成交量_万", ascending=False)
```
